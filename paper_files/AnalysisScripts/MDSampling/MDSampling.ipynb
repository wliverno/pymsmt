```python
import pytraj as pt
import numpy as np
import matplotlib.pyplot as plt
import sander

def stretchBond(atoms, dist):
    [A, B] = atoms
    axis_vector = B-A
    u = axis_vector/np.linalg.norm(axis_vector)
    return dist*u - axis_vector


```

# DT Sampling


```python
mask = ":1@N3:1@H3"
traj = pt.load('DT.pdb' , top='DT.prmtop')
opt=sander.gas_input(0)
opt.cut=9999
print(opt.__dir__())
opt.ew_type=1
print(pt.energy_decomposition(traj, mm_options=opt))
print()
print(opt.ew_type, opt.ipb)


atoms = [a for a in traj[mask][0]]
print(atoms)
unitVec = atoms[1]-atoms[0]
print(np.linalg.norm(unitVec))
unitVec /= np.linalg.norm(unitVec)
#unitVec = np.array([0,1,0])
print('UNIT VECTOR',unitVec)

t0 = pt.Trajectory(top=traj.top)
dist = np.arange(-1,2.6,0.1);
Elist2 = []
for r in dist:
    modTraj = traj.copy()
    translation = unitVec*1*r
    transMask = f':1@H3 x {translation[0]} y {translation[1]} z {translation[2]}'
    pt.translate(modTraj, transMask)
    energy = pt.energy_decomposition(modTraj, mm_options=opt)['tot']*0.04337
    print(r, energy[0])
    Elist2.append(energy[0])
    t0.append(modTraj[0])
    
pt.write_traj("DT_mov.pdb", t0, overwrite=True, options='model')
```

```python
plt.plot(dist, Elist2-min(Elist2))
plt.xlabel(r'Distance ($\AA$)')
plt.ylabel('Energy (eV)')
```

```python
def harmRest(ang, dist):
    theta = abs(180-ang)
    r = abs(dist - 2.225)
    if theta>10:
        theta = 10
    if r<0.05:
        r = 0
    elif r<0.25:
        r -= 0.05
    if r>=0.25:
        r = 0.2
    theta *= np.pi/360
    return 100*(theta**2) + 100*(r**2)
```

# AG1-O4 Angle Restrained Sampling

```python
mask = ":2@O4:3@AG"
traj = pt.load('SingleCAg2T1BP.pdb' , top='SingleCAg2T1BP_angRest.prmtop')
opt=sander.gas_input(0)
opt.cut=9999
opt.ntc=3

atoms = [a for a in traj[mask][0]]
unitVec = atoms[0]-atoms[1]
unitVec /= np.linalg.norm(unitVec)
#unitVec = np.array([0,1,0])
print('UNIT VECTOR',unitVec)

t0 = pt.Trajectory(top=traj.top)
dist = np.linspace(-0.5,0.5, 17);
print('Stretch,Etot')
Elist2 = []
for r in dist:
    modTraj = traj.copy()
    translation = unitVec*-1*r
    transMask = f':3@AG x {translation[0]} y {translation[1]} z {translation[2]}'
    pt.translate(modTraj, transMask)
    #print(modTraj[mask][0][0])
    ang = pt.angle(modTraj, ':1@N3 :4@AG :2@N3')[0]
    dist = pt.distance(modTraj, ':2@O4 :3@AG')[0]
    eDict = pt.energy_decomposition(modTraj, mm_options=opt)
    energy = (eDict['tot']+harmRest(ang,dist))*0.04337
    print(f'{r},{energy[0]}')
    Elist2.append(energy[0])
    t0.append(modTraj[0])
    
pt.write_traj("Ag1O4AngRestSample.pdb", t0, overwrite=True, options='model')
```
